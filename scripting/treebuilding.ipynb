Note: For the statistics to be meaningful, the tree generatedfor any expression must be deterministic, otherwise a single expression could have different representations for the same information. Thus, we will be imposing the following rules:

- Associativity is flattened into multiple pranches under 1 parent.
$$
(a+(b+c)) \rightarrow (a,b,c)
$$
$$
(a*(b*c)) \rightarrow * (a,b,c)
$$

- Function chains are chains of unaries
$$
sin(exp(log(x))) $$

$$
\downarrow
$$

$$
sin \rightarrow exp \rightarrow log \rightarrow x
$$

- Constant functions are evaluated
$$
2+3 \rightarrow 5
$$
$$
x*1 \rightarrow x
$$
$$
x*0 \rightarrow 0
$$

- Subtraction: Note this actually doesn't add any complexity because within any non-unary, we implicitly allow for any linear transformation. Similarly, we allow for a linear transformation in the input variables inside each leaf node, so this doesn't create any extra structure
$$
a - b = + (a, -b)
$$

- Operator simplicity: For the purposes of this paper, the operators will only come from a simplified subset. Allowed unary functions include trigonometric functions, exponentials, and polynomials. Binary operations are limited to +, /, and *. All derivatives will be considered part of the variable set and not its own function. More complex functions like Argmin, Integration, vector operations will be omitted.
